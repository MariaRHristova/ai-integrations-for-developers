```python
def analyze_document_pipeline(path_to_document: Path):
    """
    Loads a document, summarizes it, extracts keywords, and categorizes it.

    Args:
        path_to_document (Path): The path to the document file (e.g., PDF).

    Returns:
        tuple: A tuple containing (doc_summary, keywords, categorization).
    """
    print(f"Processing document: {path_to_document.name}")

    # 1. Load document (upload to Anthropic Files API)
    with open(path_to_document, "rb") as file_handle:
        uploaded_doc = a_client.beta.files.upload(
            file=(
                path_to_document.name,
                file_handle,
                "application/pdf", # Assuming PDF, adjust if other types are supported
            )
        )
    print(f"Uploaded file ID: {uploaded_doc.id}")

    # 2. Summarize the document using Anthropic
    anthropic_response = a_client.beta.messages.create(
        model=a_model,
        system="You are a helpful HR assistant that summarizes CVs",
        max_tokens=1024,
        betas=["files-api-2025-04-14"],
        messages=[
            {
                "role": "user",
                "content": [
                    {
                        "type": "text",
                        "text": "Please summarize this document.",
                    },
                    {
                        "type": "document", # Corrected: Top-level content type should be 'document'
                        "source": {
                            "type": "file",     # This type is for the source of the file_id
                            "file_id": uploaded_doc.id, # Corrected: Key for ID is 'file_id'
                        },
                    },
                ],
            }
        ],
    )
    doc_summary = anthropic_response.content[0].text
    print("\n--- Document Summary ---")
    print(doc_summary)

    # 3. Extract keywords using OpenAI
    keyword_response_obj = o_client.responses.create(
        model="gpt-5.4-nano", # Using OpenAI client for keyword extraction
        input=[
            {"role": "system", "content": "You are a helpful HR assistant that extracts keywords"},
            {"role": "user", "content": doc_summary}
        ]
    )
    keywords = keyword_response_obj.output_text
    print("\n--- Extracted Keywords ---")
    print(keywords)

    # 4. Categorize keywords using OpenRouter with structured output
    categorization_response_obj = client.responses.parse(
        model="gpt-5.4-nano", # Using OpenRouter client for categorization
        input=[
            {"role": "system", "content": "You are an expert in HR summarization of CVs. Output the confidence between 0 and 1"},
            {"role": "user", "content": f"Categorize this document based on the summary: {doc_summary}"},
        ],
        text_format=KeyWordCategory,
    )
    categorization = categorization_response_obj.output_parsed
    print("\n--- Document Categorization ---")
    print(f"Category: {categorization.category}")
    print(f"Confidence: {categorization.confidence}")

    return doc_summary, keywords, categorization
```

```python
# Assuming path_to_file is already defined from a previous cell
# For example: path_to_file = Path("/content/Mariya_Hristova_CV_Resume.pdf")

summary, extracted_keywords, categorized_output = analyze_document_pipeline(path_to_file)

print("\n\n--- Final Results ---")
print("Summary:", summary[:100] + "...") # Print first 100 chars of summary
print("Keywords:\n", extracted_keywords[:100] + "...") # Print first 100 chars of keywords
print("Categorization:", categorized_output)
```
